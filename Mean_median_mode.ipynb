```python
#Quantity

average_Quantity = np.mean(df['Quantity'])
print(f"Average Quantity of Stock purchased', {average_Quantity:.2f} units")
median_Quantity = np.median(df['Quantity'])
print(f"Median Quantity of Stock purchased', {median_Quantity:.2f} units")
mode_Quantity = df["Quantity"].mode()[0]
print(f"Mode of Quantity:', {mode_Quantity:.2f} units")
```

```python
#Total Amount

average_total_sales = np.mean(df['Total Amount'])
print(f"Average Total Sales', ${average_total_sales:.2f}")
median_total_sales = np.median(df['Total Amount'])
print(f"Median of Total Sales', ${median_total_sales:.2f}")
mode_total_sales = df["Total Amount"].mode()[0]
print(f"Mode of Total Sales:', ${mode_total_sales:.2f}")
```

```python
#Age

median_cus_age = np.median(df['Age'])
ave_cus_age = np.mean(df['Age'])
print(f"Average Quantity of Stock purchased', {ave_cus_age:.0f} years old")
print(f"Median Customer Age', {median_cus_age:.0f} years old")
mode_cus_age = df["Age"].mode()[0]
print(f"Mode of Quantity:', {mode_cus_age:.0f} years old")
```

```python
#Product Category Performance

Best_performer_dep = df["Product Category"].mode()[0]
print("Mode of Product Category:", Best_performer_dep)
#average_Category = np.mean(df['Product Category'])
#print(f"Average Product Category performance', {average_Category}")
#median_Category = np.median(df['Product Category'])
#print(f"Median Product Category Perormance', {median_Category}")
###====================================================================================================

#Product Category
# 1. Get the Mode (Most frequent category)
most_frequent_categories = df["Product Category"].mode().tolist()
category_counts = df["Product Category"].value_counts()
category_percentages = df["Product Category"].value_counts(normalize=True) * 100

# Print the results neatly
print(f"Most Popular Category (Mode): {', '.join(most_frequent_categories)}")
print("\n--- Full Breakdown ---")
for category, count in category_counts.items():
    pct = category_percentages[category]
    print(f"{category}: {count} items ({pct:.1f}%)")
```

```python
#Price per Unit
average_unit_price = np.mean(df['Price per Unit'])
print(f"Average Total Sales', ${average_unit_price:.2f}")
median_unit_prices = np.median(df['Price per Unit'])
print(f"Median of Total Sales', ${median_unit_prices:.2f}")
mode_unit_price = df["Price per Unit"].mode()[0]
print(f"Mode of Total Sales:', ${mode_unit_price:.2f}")
```

```python
def Stats():

    print("Average Total Amount:", df['Total Amount'].mean())

    print("Median:", df['Total Amount'].median())

    print("Mode:", df['Total Amount'].mode())

    print('-------------------------------------')

    print("Average Price per Unit:", df['Price per Unit'].mean())

    print("Median:", df['Price per Unit'].median())

    print("Mode:", df['Price per Unit'].mode())

    print('-------------------------------------')

    print("Average Quantity:", df['Quantity'].mean())

    print("Median:", df['Quantity'].median())

    print("Mode:", df['Quantity'].mode())
```
